# Train Data

## P_values 

```python
p_cls_train = [p_cl_train1[['Variable', 'df1_Stat_Sig']].rename(columns = {'df1_Stat_Sig': 'df1'}), 
         p_cl_train2[['Variable', 'df2_Stat_Sig']].rename(columns = {'df2_Stat_Sig': 'df2'}), 
         p_cl_train3[['Variable', 'df3_Stat_Sig']].rename(columns = {'df3_Stat_Sig': 'df3'}), 
         p_cl_train4[['Variable', 'df4_Stat_Sig']].rename(columns = {'df4_Stat_Sig': 'df4'})]

p_cl_train = reduce(lambda left, right: pd.merge(left, right, on='Variable', how='outer'), p_cls_train)
p_cl_train.index.name = 'p_cl'
p_cl_train
```

## Chi2 

```python
chi_trains = [chi_train1[['Variable', 'df1_Stat_Sig']].rename(columns = {'df1_Stat_Sig': 'df1'}), 
         chi_train2[['Variable', 'df2_Stat_Sig']].rename(columns = {'df2_Stat_Sig': 'df2'}), 
         chi_train3[['Variable', 'df3_Stat_Sig']].rename(columns = {'df3_Stat_Sig': 'df3'}), 
         chi_train4[['Variable', 'df4_Stat_Sig']].rename(columns = {'df4_Stat_Sig': 'df4'})]

chi_train = reduce(lambda left, right: pd.merge(left, right, on='Variable', how='outer'), chi_trains)
chi_train.index.name = 'Chi2'
chi_train
```

## Mann Whitney 

```python
mann_trains = [mann_train1[['Variable', 'df1_Stat_Sig']].rename(columns = {'df1_Stat_Sig': 'df1'}), 
         mann_train2[['Variable', 'df2_Stat_Sig']].rename(columns = {'df2_Stat_Sig': 'df2'}), 
         mann_train3[['Variable', 'df3_Stat_Sig']].rename(columns = {'df3_Stat_Sig': 'df3'}), 
         mann_train4[['Variable', 'df4_Stat_Sig']].rename(columns = {'df4_Stat_Sig': 'df4'})]

mann_train = reduce(lambda left, right: pd.merge(left, right, on='Variable', how='outer'), mann_trains)
mann_train.index.name = 'Mann'
mann_train
```
